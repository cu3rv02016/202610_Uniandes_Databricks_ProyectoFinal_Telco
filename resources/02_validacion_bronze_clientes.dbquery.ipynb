```sql
SELECT
    COUNT(*) AS total_clientes,
    COUNT(DISTINCT cliente_id) AS clientes_unicos,
    SUM(
        CASE
            WHEN cliente_id IS NULL THEN 1
            ELSE 0
        END
    ) AS clientes_id_nulos
FROM workspace.telco_bronze.clientes_raw;

DESCRIBE DETAIL workspace.telco_bronze.clientes_raw;
```
